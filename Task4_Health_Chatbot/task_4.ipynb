## Problem Statement
Access to reliable health information is often limited, especially
in underserved areas. In this task we build a chatbot that uses a
real LLM API to answer general health queries in a friendly,
clear, and safe manner — making basic health guidance more accessible.

## Goal
- To build a Python-based chatbot using a real LLM API (Groq/LLaMA)
- To apply prompt engineering to shape the model's tone and behavior
- To implement safety filters blocking harmful or sensitive queries
- To test the chatbot with real health questions and evaluate responses

API Setup & Prompt Engineering:

```python
from groq import Groq

GROQ_API_KEY = "your_groq_api_key_here"
client = Groq(api_key=GROQ_API_KEY)

SYSTEM_PROMPT = """You are a friendly and knowledgeable medical assistant.
Answer general health questions clearly and simply.
Always remind users to consult a real doctor for serious concerns.
Never diagnose conditions or recommend specific dosages."""

blocked_keywords = ['overdose', 'self-harm', 'suicide', 'illegal drug']

print("Setup complete!")
```

Safety Filter and Chatbot Function:

```python
def ask_health_question(user_question):
    for keyword in blocked_keywords:
        if keyword.lower() in user_question.lower():
            return 'I cannot assist with that query. Please contact a medical professional.'

    response = client.chat.completions.create(
       model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": user_question}
        ]
    )
    return response.choices[0].message.content

print("Chatbot ready!")
```

Testing Chatbot:

```python
questions = [
    'What causes a sore throat?',
    'Is paracetamol safe for children?',
    'How much water should I drink per day?',
    'What are symptoms of diabetes?'
]

for q in questions:
    print(f'Q: {q}')
    print(f'A: {ask_health_question(q)}')
    print('-' * 60)
```

## Explanation

### What Was Built:
A General Health Query Chatbot powered by LLaMA 3.1 via Groq API
that answers health questions in a friendly medical assistant tone.

### Prompt Engineering Applied:
- System prompt designed to make the model respond as a helpful
  medical assistant
- Tone kept friendly, clear, and non-diagnostic
- Model instructed to always recommend consulting a real doctor

### Safety Filters:
- Blocked keywords list prevents harmful queries from reaching the LLM
- Queries containing words like overdose, self-harm, suicide, or
  illegal drug return a safety message instead of an LLM response

### Example Results:
- Sore throat query — model explained viral and bacterial causes clearly
- Paracetamol safety — model gave age-appropriate guidance with warnings
- Water intake — model gave general recommendation with health context
- Diabetes symptoms — model listed symptoms and advised medical evaluation

### Key Takeaway:
Prompt engineering significantly shapes LLM behavior without any
model training. A well-designed system prompt transforms a general
LLM into a domain-specific assistant. Safety filters are essential
in health applications to prevent potentially harmful responses.
